## Implementation check

A small behavior check precedes the real-data measurements.


```python
from numpy_layers import BatchNorm1d,MLPWithBN
x=np.array([[1.,2.],[2.,3.],[4.,1.],[3.,4.]])
normalized=BatchNorm1d(2).forward(x)
assert np.allclose(normalized.mean(axis=0),0)
model=MLPWithBN(input_dim=2,hidden_dims=(3,));model.step(x,np.array([0,1,1,0]))
assert np.isfinite(model.forward(x,False)).all()
```
